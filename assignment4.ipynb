```python
import pandas as pd
import numpy as np

#for nominal dataset
data=pd.DataFrame({
    'color':['red','blue','red','green'],
    'city':['delhi','mumbai','delhi','kolkata']
})

print("NOMINAL DATASET")
print(data)

n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):
    count=0

    for col in data.columns:
      if data.loc[i,col] !=data.loc[j,col]:
        count += 1
    matrix[i,j] = count / len (data.columns)
print("\nDISIMILARITY MATRIX - NOMINAL")
print(pd.DataFrame(
    matrix,
    index=['01','02','03','04'],
    columns=['01','02','03','04']
).round(2))
```

```python
import pandas as pd
import numpy as np

#Binary dataset
data=pd.DataFrame({
    'student':[1,1,0,0],
    'Employed':[0,1,1,0],
    'purchased':[1,1,0,1]
})

print("BINARY DATASET")
print(data)

n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):

    diff=0

    for col in data.columns:
      if data.loc[i,col] !=data.loc[j,col]:
        diff += 1
    matrix[i,j] = diff / len(data.columns)
print("\nDISIMILARITY MATRIX - NOMINAL")
print(pd.DataFrame(
    matrix,
    index=['01','02','03','04'],
    columns=['01','02','03','04']
).round(2))

```

```python
import pandas as pd
import numpy as np

#for ordrinal dataset
data=pd.DataFrame({
   'satisfaction': ['low','medium','high','medium']
})
print("ORDINAL DATASET")
print(data)

#convert ordinal values into ranks
rank={
    'low':0,
    'medium':1,
    'high':2
}
n=len(data)
matrix=np.zeros((n,n))

for i in range(n):
  for j in range(n):

    x=rank[data.loc[i,'satisfaction']]
    y=rank[data.loc[j,'satisfaction']]

    #ordinal dissimilarity
    matrix[i,j]=abs(x-y)/2
print("\nDISIMILARITY MATRIX - ORDINAL")
print(pd.DataFrame(
    matrix,
    index=['01','02','03','04'],
    columns=['01','02','03','04']
).round(2))
```
