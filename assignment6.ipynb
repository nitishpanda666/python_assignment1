```python
import pandas as pd
import numpy as np
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
data=pd.DataFrame({
    'age':[15,16,17,18,19,20],
    'studytime':[1,2,3,2,4,3],
    'failures' :[0,1,0,2,1,0]
}).astype(float)
```

```python
minimax_scalar = MinMaxScaler()
minmax_scaled = minimax_scalar.fit_transform(data)
```

```python
standard_scalar = StandardScaler()
zscore_scaled = standard_scalar.fit_transform(data)
```

```python
j=np.ceil(np.log10(data.abs().max()))
decimal_scaled = data / (10 ** j)
```

```python
print("\n=====ORIGINAL DATA======")
print(data)
```

```python
print("\n=====MIN-MAX SCALED DATA======")
print(pd.DataFrame(minmax_scaled, columns=data.columns))
```

```python
print("\n=====Z-SCORE SCALED DATA======")
print(pd.DataFrame(zscore_scaled, columns=data.columns))
```

```python
print("\n=====DECIMAL SCALED DATA======")
print(decimal_scaled)
```
